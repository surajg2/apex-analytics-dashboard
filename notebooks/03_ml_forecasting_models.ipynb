## 🎯 Objective  
Trains supervised machine learning time-series models (Gradient Boosting, Random Forest, Ridge) on historical sales data and evaluates holdout forecasting performance.

```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sys.path.append("..")
from utils.forecasting_engine import SalesForecaster

daily_df = pd.read_csv("../data/processed/daily_sales_series.csv")
daily_df["date"] = pd.to_datetime(daily_df["date"])
print(f"Loaded {len(daily_df):,} daily sales observations.")
daily_df.tail()
```

## 1. Train Gradient Boosting Forecaster & Evaluate Metrics

```python
forecaster = SalesForecaster(model_type="gradient_boosting")
eval_res = forecaster.train_and_eval(daily_df, test_days=30)

print("=== Forecast Metrics ===")
for metric, val in eval_res["metrics"].items():
    print(f"{metric}: {val}")
```

## 2. Generate 60-Day Future Revenue Forecast

```python
future_df = forecaster.predict_future(daily_df, forecast_horizon_days=60)
future_df.head(10)
```

```python
plt.figure(figsize=(14, 6))
plt.plot(daily_df["date"].tail(90), daily_df["revenue"].tail(90), label="Historical Sales", color="#38bdf8", lw=2)
plt.plot(future_df["date"], future_df["forecast_revenue"], label="60-Day Forecast", color="#818cf8", lw=2, linestyle="--")
plt.fill_between(future_df["date"], future_df["lower_bound"], future_df["upper_bound"], color="#818cf8", alpha=0.2, label="95% Confidence Interval")
plt.title("E-Commerce Daily Revenue 60-Day ML Forecast", fontsize=14, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Revenue ($)")
plt.legend()
plt.show()
```
